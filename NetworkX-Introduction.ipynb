# NetworkX Introduction: Nodes and Edges

This notebook introduces network graphs by building a very small research collaboration network one step at a time.

We will:

- create an empty graph
- add researchers as nodes
- store information as node attributes
- connect researchers with edges
- visualize the network
- inspect neighbors, degree, and connectivity

The goal is to understand the basic graph model before building networks automatically from data.


## 1. Setup

We will use NetworkX to create the graph and Matplotlib to display it.


````python
import networkx as nx
import matplotlib.pyplot as plt
````

## 2. Create an Empty Graph

`nx.Graph()` creates an undirected graph. An undirected edge represents a relationship without a direction.


````python
G = nx.Graph()

print("Nodes:", list(G.nodes()))
print("Edges:", list(G.edges()))
````

## 3. Add Researchers as Nodes

A node represents an entity in the network. Here, each node represents a researcher.

The node ID is a short name used by NetworkX. Additional information can be stored as attributes.


````python
G.add_node(
    "jane",
    name="Jane Park",
    field="Transportation Engineering"
)

G.add_node(
    "john",
    name="John Rivera",
    field="Mathematical Optimization"
)

G.nodes(data=True)
````

## 4. Add a Collaboration as an Edge

An edge represents a relationship between two nodes.

Jane and John have collaborated, so we connect them.


````python
G.add_edge("jane", "john")

print("Nodes:", list(G.nodes()))
print("Edges:", list(G.edges()))
````

## 5. Add Two More Researchers

Priya has collaborated with Jane and John. Carlos has collaborated with Priya.


````python
G.add_node(
    "priya",
    name="Priya Shah",
    field="Mobility Systems"
)

G.add_node(
    "carlos",
    name="Carlos Mendes",
    field="Operations Research"
)

G.add_edge("priya", "jane")
G.add_edge("priya", "john")
G.add_edge("carlos", "priya")

print("Nodes:", G.number_of_nodes())
print("Edges:", G.number_of_edges())
````

## 6. Visualize the Network

A layout algorithm assigns positions to the nodes for display. The positions are part of the visualization, not the underlying graph data.


````python
pos = nx.spring_layout(G, seed=42)

labels = {
    node: data["name"]
    for node, data in G.nodes(data=True)
}

plt.figure(figsize=(8, 6))

nx.draw(
    G,
    pos,
    labels=labels,
    with_labels=True,
    node_size=1800,
    font_size=10
)

plt.title("Research Collaboration Network")
plt.axis("off")
plt.show()
````

## 7. Inspect Relationships

NetworkX lets us ask simple questions about the graph.

Who is directly connected to Priya?


````python
list(G.neighbors("priya"))
````

How many direct connections does each researcher have?

This is the node's degree.


````python
for node, degree in G.degree():
    print(
        G.nodes[node]["name"],
        degree
    )
````

Is the graph connected?

A connected graph has a path between every pair of nodes.


````python
nx.is_connected(G)
````

## 8. What We Built

The network contains:

- nodes representing researchers
- node attributes describing researchers
- edges representing collaborations

The basic pattern is:

```text
create a graph
    ↓
add nodes
    ↓
add edges
    ↓
analyze and visualize
```
